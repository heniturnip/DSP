## Memahami persamaan sinyal

### Sinyal Sinusoidal

**Fungsi Sinyal Sinusiodal**
$$
\text(signal = sin{\theta})
$$

- Fungsi sinus adalah fungsi matematika yang menghasilkan nilai antara -1 dan 1. Fungsi ini sering digunakan untuk menggambarkan gelombang periodik, seperti gelombang suara atau gelombang cahaya. (\theta):

**Fungsi Theta**
$$
{\theta}={2np.phift} 
$$

pertama - tama buat time axis untuk sumbu x

```python
time_axis = np.linspace(0, 1, 1000)
#print(time_axis)
```

```python
frekuensi = 5

signal = np.sin(2 * np.pi * frekuensi * time_axis)
#print(signal)
```

mari kita visualisasikan sinyal sinusoidal ini.

```python
plt.figure(figsize=(8, 4))
plt.plot(time_axis, signal)

# supaya grafik lebih rapi, kita lakukan hal-hal berikut
plt.title('Sinyal Sinusoidal')
plt.xlabel('Waktu (s)')
plt.ylabel('Amplitudo')
plt.grid(True)
```

### Membuktikan Representasi Sinyal Komplex dengan Python

**Definisi Sinyal Kompleks:**

$$
\text{coplex-signal}= e^{j\theta}
$$

$$
\text{coplex-signal}= e^{j2np.phift}
$$


```python
time_axis = np.linspace(0, 1, 1000)
frekuensi = 5
complex_signal = np.exp(1j * 2 * np.pi * frekuensi * time_axis)
print(complex_signal)
```

complex signal terdiri dari:
- Bagian real $cos({2np.phift})$
- Bagian imajiner : $sin({2np.phift})$

```python
bagian_real = np.real(complex_signal)
bagian_imaginer = np.imag(complex_signal)

print(f"Bagian Real : {bagian_real[3]}")
print(f"Bagian Imaginer : {bagian_imaginer[3]}")
```

```python
# plot in subplot

ax, fig = plt.subplots(2, 1, figsize=(8, 4))
fig[0].plot(time_axis, bagian_real)
fig[0].set_title('Bagian Real (Cosinus)')
fig[0].grid(True)
fig[0].set_xlabel('Waktu (s)')
fig[0].set_ylabel('Amplitudo')
fig[1].plot(time_axis, bagian_imaginer)
fig[1].set_title('Bagian Imaginer (Sinus)')
fig[1].grid(True)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10, 4))
plt.plot(time_axis, bagian_real, label='Bagian Real (cosinus)')
plt.plot(time_axis, bagian_imaginer, label='Bagian Imaginer (sinus)')
plt.title('Bagian Real dan Imaginer')
plt.xlabel('Waktu (s)')
plt.ylabel('Amplitudo')
plt.grid
plt.legend(loc = 'lower right')
plt.tight_layout()
plt.show()
```

```python
magnitude = np.abs(complex_signal)

plt.figure(figsize=(10, 2))
plt.plot(time_axis, magnitude)
plt.title('Magnitude')
plt.xlabel('Waktu (s)')
plt.ylabel('Magnitude')
plt.grid(True)
plt.show()
```

```python
phase = np.angle(complex_signal)

plt.figure(figsize=(10, 2))
plt.plot(time_axis, complex_signal)
plt.plot(time_axis, phase)
plt.title('Phase Sinyal')
plt.xlabel('Waktu (s)')
plt.ylabel('Amplitudo')
plt.grid(True)
plt.show()
```
