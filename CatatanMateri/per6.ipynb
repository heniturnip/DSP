# Membuktikan prinsip Proporsionality

Dari hasil grafik dibawah ini, terbukti bahwa sinyal jika dilakukan scaling dahulu lalu masuk ke sistem akan sama dengan sinyal yang dimasukan sistem dulu baru dilakukan scaling

```python
import numpy as np
import matplotlib.pyplot as plt

# Membuat fungsi sistem linier
def linear_system(input_signal):
    return 2*input_signal

time_axis = np.linspace(0, 2* np.pi, 1000)
input_signal1 = np.sin(time_axis)

# Melakukan Scalling dulu baru di input ke sistem
scaling = 3
is1 = input_signal1 * scaling
output_signal1 = linear_system(is1)
print(f'Sebelum di scaling : {input_signal1[2]}')
print(f'Sesudah di scaling : {is1[2]}')
print(f'Output signal : {output_signal1[2]}')

# Masukkan ke sistem dulu baru di scaling
output_signal2 = linear_system(input_signal1)
os2 = output_signal2 * scaling
print(f'Output Sinyal sebelum di scaling : {output_signal2[2]}')
print(f'Output Sinyal Sesudah di scaling : {os2[2]}')

# Plotting
plt.figure()
plt.plot(time_axis, input_signal1, label='Input Signal 1', color='Green')
plt.plot(time_axis, output_signal1, label='Output Signal 1', color='orange')
plt.plot(time_axis, os2, label='Output Signal 2', color='purple', linestyle='--')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.title('Proporsionality')
plt.grid()
plt.legend()
plt.show()
```

# Membuktikan prinsip Superpropotion

Dari hasil visualisasi dibawah ini terbukti bahwa sinyal jika dijumlahkan lalu dimasukan kesistem akan sama jika sinyal dimasukan kesistem baru dijumlahkan

```python
import numpy as np
import matplotlib.pyplot as plt

# Membuat fungsi sistem linier
def linear_system(input_signal):
    return 2*input_signal

time_axis = np.linspace(0, 2* np.pi, 1000)
input_signal1 = np.sin(time_axis)
input_signal2 = np.cos(time_axis)

# Menjumlahkan sinyal 1 dan sinyal 2 lalu masuk ke sistem
input_signal3 = input_signal1 + input_signal2
output_signal3 = linear_system(input_signal3)
print(f'input signal 3 : {input_signal3[2]}')
print(f'Output Signal 3 : {output_signal3[2]}')

# Memasukkan sinyal 1 dan sinyal 2 ke sistem lalu dijumlahkan
output_signal4 = linear_system(input_signal1)
output_signal5 = linear_system(input_signal2)
jumlah = output_signal4 + output_signal5
print(f'Output Signal 4 : {output_signal4[2]}')
print(f'Output Signal 5 : {output_signal5[2]}')
print(f'Jumlah : {jumlah[2]}')

# Plotting
plt.figure()
plt.plot(time_axis, output_signal3, label='Output Signal 3', color='orange')
plt.plot(time_axis, jumlah, label='Jumlah', color='purple', linestyle='--')
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.title('Superposition')
plt.grid()
plt.legend()
plt.show()
```
